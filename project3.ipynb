The main idea of using N-grams for text generation is to assume that the last word <tt>x<sub>n</sub></tt> of the n-gram can be inferred from the other words that appear in the same n-gram <tt>x<sub>n-1</sub>, x<sub>n-2</sub>, …, x<sub>1</sub></tt>. We will call these sequences of preceeding words a <tt>history</tt>. With this simplification, in order to predict the next word, we just need to look back for <tt>n-1</tt> tokens. 

Assume that we want to predict the most likely word to complete the following sentence:

<tt>I went to ___</tt>

The N-gram model will pick the next word 'semi-randomly', such that the words that have a higher probability will have higher chances of being produced, while other words with lower probability have a chance of being generated. In a bigram model, for example, if <tt>p(class|to)</tt> is greater than <tt>p(library|to)</tt>, the model will generate <tt>I went to class</tt>.

As we mentioned before, to predict a word at position <tt>n</tt>, we would have to look at all previous <tt>n-1</tt> words of the N-gram model. Therefore, <tt>get_ngrams()</tt> function should return a list of tuples <tt>((history), current_word)</tt>. These tuples are computed from the tokenized string passed as an argument to the function. The function must work for any N, passed as an argument. Thus, for N=3 (trigram model) and a tokenized string <tt>[I, bought, a, red, car]</tt>, the list of tuples is: 

<tt>
[((START_TOKEN, START_TOKEN), I),
((START_TOKEN, I), bought),
((I, bought), a),
((bought, a), red),
((a, red), car)]
</tt>


Note that <tt>START_TOKEN</tt> is a special symbol. When we need to generate the first token of a sentence, there is no preceeding history and to address this issue, we will introduce the leading tags, <START_TOKEN> to make sure that at any point in text generation, we are always working with correct N-grams.
